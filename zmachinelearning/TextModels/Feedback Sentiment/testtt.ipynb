```python
import joblib
import pandas as pd

# Load the saved model and vectorizer
model = joblib.load('sentiment_model.pkl')
tfidf = joblib.load('tfidf_vectorizer.pkl')

# Preprocess the input text (same preprocessing as in the training phase)
def preprocess_feedback(feedback_text):
    feedback_text = feedback_text.lower()  # Convert to lowercase
    return feedback_text

# Function to predict sentiment of a new review
def predict_sentiment(feedback):
    processed_feedback = preprocess_feedback(feedback)
    
    # Transform the feedback using the same vectorizer
    feedback_tfidf = tfidf.transform([processed_feedback])
    
    # Predict sentiment using the trained model
    sentiment_probabilities = model.predict_proba(feedback_tfidf)[0]
    sentiment_score = sentiment_probabilities[1]  # Probability of positive sentiment
    sentiment_label = "Positive" if sentiment_score > 0.5 else "Negative"
    
    return sentiment_label, sentiment_score

# Test the model with new input
def test_model():
    # Sample reviews (You can replace these with any input text)
    reviews = [
        "The room was fantastic, I had a great stay!",
        "The service was terrible, I would never come back again.",
        "It was okay, nothing special but fine for the price."
    ]
    
    for review in reviews:
        sentiment_label, sentiment_score = predict_sentiment(review)
        print(f"Review: {review}")
        print(f"Predicted Sentiment: {sentiment_label} (Score: {sentiment_score:.2f})\n")

# Run the test script
if __name__ == "__main__":
    test_model()
```
