```python
import pickle
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder

# Load model, vectorizer, and label encoder
with open('complaint_category_model.pkl', 'rb') as model_file:
    model = pickle.load(model_file)

with open('tfidf_vectorizer.pkl', 'rb') as tfidf_file:
    tfidf = pickle.load(tfidf_file)

with open('label_encoder.pkl', 'rb') as le_file:
    label_encoder = pickle.load(le_file)


# Function to preprocess and predict category for new complaint
def predict_category(new_complaint):
    processed_text = new_complaint  # Use same preprocess_text function
    tfidf_features = tfidf.transform([processed_text])
    prediction = model.predict(tfidf_features)
    category = label_encoder.inverse_transform(prediction)
    return category[0]

# Example usage
new_complaint = "Two person were fighting with each other"
predicted_category = predict_category(new_complaint)
print("Predicted Category:", predicted_category)
```
