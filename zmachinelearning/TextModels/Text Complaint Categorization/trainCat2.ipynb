```python
import pandas as pd
import torch
import torch.nn as nn
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report
```

```python
data = pd.read_csv("railma.csv")
```

```python
label_encoder_type = LabelEncoder()
label_encoder_subtype = LabelEncoder()
data['complaint_type'] = label_encoder_type.fit_transform(data['complaint_type'])
data['sub_type'] = label_encoder_subtype.fit_transform(data['sub_type'])
```

```python
# TF-IDF Vectorization
tfidf = TfidfVectorizer(max_features=5000)
X = tfidf.fit_transform(data['complaint']).toarray()
```

```python
# Split data
y_type = data['complaint_type']
y_subtype = data['sub_type']
X_train, X_test, y_train_type, y_test_type = train_test_split(X, y_type, test_size=0.2, random_state=42)
_, _, y_train_subtype, y_test_subtype = train_test_split(X, y_subtype, test_size=0.2, random_state=42)
```
