```python
import random

# List of random names
names = ["Ashok Kumar", "Rajesh Singh", "Suresh Reddy", "Vikram Yadav", "Santosh Sharma", 
         "Neha Gupta", "Rajan Verma", "Kavita Bhat", "Amit Shah", "Shivani Singh", "Shashank Yadav", "Talha Ansari", "Atharva Bansod"]

# List of random genders
genders = ["Male", "Female", "Other"]

# List of berths
berths = ["Lower", "Middle", "Upper"]

# List of coach positions
coach_positions = ["A1", "B1", "C1", "D1", "E1", "F1", "G1", "H1", "J1", "K1"]

# Generate a random seat number (e.g., 1 to 60 for each coach)
def generate_seat_number():
    return random.randint(1, 60)

def get_train_details(pnr):
    # Generate random data
    name = random.choice(names)
    age = random.randint(18, 70)
    gender = random.choice(genders)
    berth = random.choice(berths)
    coach_position = random.choice(coach_positions)
    seat_number = generate_seat_number()  # Random seat number

    # Train number is always 12480
    train_no = 12480
    
    # Output the details
    return {
        "Train Number": train_no,
        "PNR": pnr,
        "Passenger Name": name,
        "Age": age,
        "Gender": gender,
        "Berth": berth,
        "Coach Position": coach_position,
        "Seat Number": seat_number
    }

# Take input of PNR number
pnr_input = input("Enter the PNR number: ")

# Get the train details for the given PNR
train_details = get_train_details(pnr_input)

# Print the output
for key, value in train_details.items():
    print(f"{key}: {value}")
```
