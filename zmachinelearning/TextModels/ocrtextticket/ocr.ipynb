```python
from paddleocr import PaddleOCR
from PIL import Image
import re

# Open the image
image_path = 'ticketB.jpg'
image = Image.open(image_path)

# Downscale the image to reduce resolution (optional)
image = image.resize((image.width // 2, image.height // 2))  # Reducing size by half

# Convert the image to RGB before saving (to avoid the RGBA error)
image = image.convert('RGB')

# Save the downscaled image in RGB mode
image.save('downscaled_image.jpg')

# Initialize PaddleOCR with English language model
ocr = PaddleOCR(use_gpu=True, lang='en')  # Enable GPU acceleration and set language to English

# Perform OCR on the downscaled image with text detection skipped (optional)
results = ocr.ocr('downscaled_image.jpg', cls=False)  # cls=False to skip text detection

# Prepare the text
all_text = [line[1][0] for line in results[0]]
combined_text = " ".join(all_text)

# Display results
print("Detected Text in One Paragraph:")
print(combined_text)

# Regular expressions for Train No, PNR No, and any number with 5 digits
train_no_regex = r'\b\d{4,5}\b'  # Matches any 5-digit number
pnr_no_regex = r'\b\d{3}-\d{7}\b'  # Matches PNR format like XXX-XXXXXXX
eight_digit_number_regex = r'\b\d{8}\b'  # Matches any 8-digit number

# Function to extract required information
def extract_info(text):
    train_numbers = re.findall(train_no_regex, text)
    pnr_numbers = re.findall(pnr_no_regex, text)
    eight_digit_numbers = re.findall(eight_digit_number_regex, text)

    # Check if lists are empty, then return a default value
    return {
        "Train Number": train_numbers[0] if train_numbers else "Not Found",
        "PNR Numbers": pnr_numbers[0] if pnr_numbers else "Not Found",
        "Ticket Number": eight_digit_numbers[0] if eight_digit_numbers else "Not Found"
    }

# Extracting the necessary information from the combined text
extracted_info = extract_info(combined_text)

# Display the extracted information
print("\nExtracted Information:")
for category, value in extracted_info.items():
    print(f"{category}: {value}")
```
